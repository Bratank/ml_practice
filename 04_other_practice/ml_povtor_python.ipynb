### Задача 3: Расчет матрицы ошибок (Confusion Matrix) и Точности (Precision / Recall)
На собеседованиях обожают просить посчитать метрики бинарной классификации без `scikit-learn`.

**Задание:** Напиши функцию `binary_metrics(y_true, y_pred_probs, threshold=0.5)`, которая:
1. Переводит вероятности `y_pred_probs` в бинарные предсказания (`1`, если `prob >= threshold`, иначе `0`).
2. Считает количество:
   * `TP` (True Positive: истинный `1` и предсказанный `1`)
   * `FP` (False Positive: истинный `0`, но предсказанный `1`)
   * `FN` (False Negative: истинный `1`, но предсказанный `0`)
3. Возвращает словарь `{"TP": tp, "FP": fp, "FN": fn, "precision": ..., "recall": ...}`, где:
   * $\text{Precision} = \frac{TP}{TP + FP}$ (если знаменатель равен 0, вернуть `0.0`)
   * $\text{Recall} = \frac{TP}{TP + FN}$ (если знаменатель равен 0, вернуть `0.0`)
   * Значения `precision` и `recall` округли до **4 знаков**.

````python
def binary_metrics(y_true: list[int], y_pred_probs: list[float], threshold: float = 0.5) -> dict:
    # ТВОЙ КОД ЗДЕСЬ
    mask = []
    for i in y_pred_probs:
        if i >=threshold:
            mask.append(1.0)
        else:
            mask.append(0.0)
    tp = 0
    fp = 0
    fn = 0
    tn = 0
    for y,a in zip(y_true,mask):
        if y and a:
            tp+=1
        elif y == 0 and a == 1:
            fp += 1
        elif y == 1 and a == 0:
            fn+=1
        else:
            tn += 1

    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    f1_score = 2 * precision * recall / (precision + recall)
    return {"TP": tp, "FP": fp, "FN": fn, "precision": precision, "recall": recall}


# === ТЕСТЫ ===
y_t = [1, 0, 1, 1, 0, 1, 0, 0]
y_p = [0.9, 0.2, 0.8, 0.4, 0.6, 0.7, 0.1, 0.3]
# Предсказания при 0.5: [1, 0, 1, 0, 1, 1, 0, 0]
# TP = 3 (индексы 0, 2, 5), FP = 1 (индекс 4), FN = 1 (индекс 3)
res = binary_metrics(y_t, y_p, threshold=0.5)
expected = {"TP": 3, "FP": 1, "FN": 1, "precision": 0.75, "recall": 0.75}
assert res == expected, f"Ошибка! Ожидалось {expected}, получено {res}"
print("✅ Задача 3 решена верно!")
````

### Задача 4: Построение словаря токенов (Vocabulary Builder)
В NLP перед подачей текста в модель слова переводят в целочисленные индексы, отсекая слишком редкие слова (шум и опечатки).

**Задание:** Напиши функцию `build_vocab(texts, min_freq=2)`, которая:
1. Принимает список строк `texts` (слова разделены пробелами), приводит все слова к нижнему регистру (`.lower()`).
2. Считает частоту каждого слова во всем корпусе.
3. Формирует словарь `word2idx`, который всегда начинается с двух специальных токенов: `{"<PAD>": 0, "<UNK>": 1}`.
4. Добавляет в этот словарь только те слова, которые встретились **не менее `min_freq` раз**. Чтобы порядок индексов был однозначным, отсортируй отобранные слова **по алфавиту** перед присвоением им индексов (`2, 3, 4...`).

````python
texts = [
    "Machine learning is great",
    "Deep learning is fun",
    "Machine learning models"
]

words = {"<PAD>": 0, "<UNK>": 1}

for s in texts:
    s = s.lower().split()
    print(s)
    for word in s:
        if word in words:
            words[word] += 1
        else:
            words[word] = 1
    print(words)
````

````python
def build_vocab(texts: list[str], min_freq: int = 2) -> dict[str, int]:
    # ТВОЙ КОД ЗДЕСЬ
    words = {}

    for s in texts:
        s = s.lower().split()

        for word in s:
            if word in words:
                words[word] += 1
            else:
                words[word] = 1

    dick = words.copy()
    for key, value in dick.items():
        if value < min_freq:
            del words[key]

    res = sorted(words.keys())
    d = ["<PAD>","<UNK>"]
    res = d + res
    for i in range(len(res)):
        res[i] = (res[i],i)
    return dict(res)
    
    

# === ТЕСТЫ ===
corpus = [
    "Machine learning is great",
    "Deep learning is fun",
    "Machine learning models"
]
# Частоты: learning (3), machine (2), is (2), great (1), deep (1), fun (1), models (1)
# Проходят порог min_freq=2 по алфавиту: 'is', 'learning', 'machine'
vocab = build_vocab(corpus, min_freq=2)
expected_vocab = {"<PAD>": 0, "<UNK>": 1, "is": 2, "learning": 3, "machine": 4}
assert vocab == expected_vocab, f"Ошибка! Ожидалось {expected_vocab}, получено {vocab}"
print("✅ Задача 4 решена верно!")
````

### Задача 5: Поиск утечки данных (Data Leakage Detector)
Критическая ошибка в ML — когда одни и те же пользователи или объекты попадают и в обучающую (`train`), и в тестовую (`test`) выборку. Кроме того, важно проверить метки классов: вдруг в тесте появился новый класс, которого не было в обучении.

**Задание:** Напиши функцию `check_data_leakage(train_ids, test_ids, train_labels, test_labels)`, которая возвращает словарь с двумя ключами:
1. `"leaked_ids"` — отсортированный по возрастанию список ID, которые одновременно присутствуют и в `train_ids`, и в `test_ids`.
2. `"unseen_classes"` — отсортированный по возрастанию список классов, которые есть в `test_labels`, но отсутствуют в `train_labels`.

````python
train_ids=[101, 102, 103, 104, 105]
test_ids=[104, 106, 101, 107]
train_labels=["cat", "dog", "cat"]
test_labels=["dog", "bird", "fish"]

train_ids = set(train_ids)
test_ids = set(test_ids)
train_labels = set(train_labels)
test_labels = set(test_labels)

print(train_ids,test_ids ,train_labels, test_labels)
````

````python
def check_data_leakage(train_ids: list, test_ids: list, train_labels: list, test_labels: list) -> dict:
    # ТВОЙ КОД ЗДЕСЬ
    train_ids = set(train_ids)
    test_ids = set(test_ids)
    train_labels = set(train_labels)
    test_labels = set(test_labels)

    leaked_id = train_ids & test_ids
    unseen_classes = (test_labels | train_labels) -  train_labels
    return {"leaked_ids" : sorted(list(leaked_id)), "unseen_classes": sorted(list(unseen_classes))}

# === ТЕСТЫ ===
res = check_data_leakage(
    train_ids=[101, 102, 103, 104, 105],
    test_ids=[104, 106, 101, 107],
    train_labels=["cat", "dog", "cat"],
    test_labels=["dog", "bird", "fish"]
)
assert res == {"leaked_ids": [101, 104], "unseen_classes": ["bird", "fish"]}, f"Ошибка! Получено: {res}"
print("✅ Задача 5 решена верно!")
````

### Задача 6: Импутация пропусков и One-Hot Encoding
Реализуй две короткие функции в **одну строчку** (через comprehensions):
1. `impute_missing(values)` — заменяет все `None` в списке чисел на среднее арифметическое всех **не-None** элементов этого же списка.
2. `one_hot_encode(label, num_classes)` — возвращает список длины `num_classes` из нулей, где на позиции `label` стоит `1`.

````python
def impute_missing(values: list[float | None]) -> list[float]:
    # ТВОЙ КОД ЗДЕСЬ (найди среднее не-None элементов и замени None через list comprehension)
    return [sum([i for i in values if i!= None])/len([j for j in values if j!= None]) if x == None else x for x in values]

def one_hot_encode(label: int, num_classes: int) -> list[int]:
    # ТВОЙ КОД ЗДЕСЬ
    return [1 if i == label else 0 for i in range(num_classes)]


# === ТЕСТЫ ===
assert impute_missing([10.0, None, 20.0, None, 30.0]) == [10.0, 20.0, 20.0, 20.0, 30.0], "Ошибка в impute_missing!"
assert one_hot_encode(2, 5) == [0, 0, 1, 0, 0], "Ошибка в one_hot_encode!"
assert one_hot_encode(0, 3) == [1, 0, 0], "Ошибка в one_hot_encode!"
print("✅ Задача 6 решена верно!")
````

Функция должна:

2. Отсортировать точки по возрастанию расстояния и взять `k` ближайших.

````python
print(1//2)
````

````python

def knn_predict(train_data: list[tuple[list[float], int]], query_point: list[float], k: int = 3) -> int:
    # ТВОЙ КОД ЗДЕСЬ
    dist = []
    for point, clas in train_data:
        dist.append((sum((a - b) ** 2 for a, b in zip(point, query_point))**0.5,clas))
        
    dist_k = sorted(dist,key = lambda x: x[0])[:k]
    k_labels = [label for dis, label in dist_k]
    return max(set(k_labels), key=k_labels.count)


# === ТЕСТЫ ===
dataset = [
    ([1.0, 1.0], 0),
    ([1.5, 2.0], 0),
    ([3.0, 4.0], 1),
    ([5.0, 7.0], 1),
    ([3.5, 5.0], 1),
    ([4.5, 5.0], 1)
]
assert knn_predict(dataset, [1.2, 1.5], k=3) == 0, "Ожидался класс 0!"
assert knn_predict(dataset, [4.0, 4.5], k=3) == 1, "Ожидался класс 1!"
print("✅ Задача 7 решена верно!")
````

### Задача 9: Декоратор кэширования инференса (Memoization)
Если ML-модель тяжелая, а пользователи часто отправляют одинаковые запросы, имеет смысл кэшировать ответы.

**Задание:** Напиши декоратор `cache_predictions(func)`, который сохраняет результаты вызовов функции `func` в словаре. 
* Если функция вызывается с аргументом `x`, который уже был в кэше — декоратор возвращает сохраненный результат **без повторного вызова** `func`.
* Если аргумента `x` в кэше нет — вызывает `func(x)`, сохраняет в кэш и возвращает ответ.

````python
def cache_predictions(func):
    # ТВОЙ КОД ЗДЕСЬ
    dick = {}
    def wrapper(*args, **kwargs):
        
        key = (args, tuple(sorted(kwargs.items())))
        if key in dick:
            return dick[key]
        dick[key] = func(*args,**kwargs)
        
        return dick[key]
    return wrapper




# === ТЕСТЫ ===
call_count = 0

@cache_predictions
def mock_heavy_model(query: str) -> int:
    global call_count
    call_count += 1
    return len(query) * 10

assert mock_heavy_model("hello") == 50
assert mock_heavy_model("hello") == 50  # Должно взяться из кэша!
assert call_count == 1, f"Модель вызвалась {call_count} раз(а) вместо 1! Кэш не сработал."
assert mock_heavy_model("ml") == 20
assert call_count == 2
print("✅ Задача 9 решена верно!")
````

**Задание:** Допиши методы класса `MinMaxScaler`:
1. `fit(self, data)` — находит минимум и максимум списка `data`, сохраняет в `self.min_` и `self.max_` и возвращает `self`.
2. `transform(self, data)` — возвращает новый список нормализованных значений (округленных до **4 знаков**). Если `self.max_ == self.min_` (все числа одинаковые), возвращает список нулей `[0.0, ...]`.
3. `fit_transform(self, data)` — последовательно вызывает `fit` и `transform`.

````python
# ТВОЙ КОД ЗДЕСЬ
class MinMaxScaler:
    def __init__(self):
        self.min_ = None
        self.max_ = None

    def fit(self, data: list[float]):
        self.min_ = min(data)
        self.max_ = max(data)
        return self

    def transform(self, data: list[float]) -> list[float]:
        dif = self.max_ - self.min_
        if dif == 0:
            return [0.0] * len(data)
        return [round((x-self.min_) / dif,4) for x in data]

    def fit_transform(self, data: list[float]) -> list[float]:
        self.fit(data)
        return self.transform(data)


# === ТЕСТЫ ===
scaler = MinMaxScaler()
train_vals = [10.0, 20.0, 30.0, 50.0]
scaled_train = scaler.fit_transform(train_vals)
assert scaled_train == [0.0, 0.25, 0.5, 1.0], f"Ошибка на train! Получено: {scaled_train}"

# Проверяем, что на тесте используются min и max от train (10 и 50)!
scaled_test = scaler.transform([30.0, 70.0])
assert scaled_test == [0.5, 1.5], f"Ошибка на test! Получено: {scaled_test}"

# Проверка деления на ноль при константном признаке
const_scaler = MinMaxScaler()
assert const_scaler.fit_transform([5.0, 5.0, 5.0]) == [0.0, 0.0, 0.0], "Ошибка на константном списке!"
print("🎉 ПОЗДРАВЛЯЮ! Все 10 задач тренажера успешно решены!")
````
