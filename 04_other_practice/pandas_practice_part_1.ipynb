## Модуль 1. Создание `DataFrame`, `Series` и управление индексами

### 🎯 Методы, которые нужны для задачи:
* `pd.DataFrame(data)` — создает двумерную таблицу из словаря списков (ключи становятся названиями колонок), списка словарей или 2D-массива NumPy.
* `df.set_index(col_name)` — делает выбранную колонку новым индексом таблицы (возвращает новый `DataFrame`, если не указан `inplace=True`).
* `df.reset_index(drop=False)` — сбрасывает текущий индекс обратно в обычную колонку и ставит стандартный числовой индекс `0, 1, ..., N-1`. Если указать `drop=True`, старый индекс просто удаляется.

### 📚 Связанные методы и атрибуты в копилку:
* `pd.Series(data, index=...)` — одномерный массив с метками (индексом). Каждая колонка `DataFrame` — это `Series`.
* `df.columns` — список (Index) названий колонок; `df.index` — индекс строк.
* `df.shape` — кортеж `(число_строк, число_колонок)`.
* `df.dtypes` — типы данных каждой колонки.
* `df.head(n)` / `df.tail(n)` — первые / последние `n` строк таблицы.
* `df.to_numpy()` — превращает `DataFrame` в чистый массив NumPy.

### Задача 1
Напиши функцию `build_users_df(ids, names, scores)`, которая:
1. Принимает три списка одинаковой длины: `ids`, `names` и `scores`.
2. Создает из них `DataFrame` с колонками `"user_id"`, `"name"` и `"score"`.
3. Делает колонку `"user_id"` индексом этой таблицы и возвращает полученный `DataFrame`.

## Модуль 2. Индексация: `.loc` против `.iloc`

### 🎯 Главная разница, которую всегда спрашивают:
* **`df.iloc[строки, столбцы]`** — индексация **по целым позициям (Integer Location)** от `0` до `N-1` (как в NumPy и списках Python).
  * В срезах `df.iloc[0:3]` правая граница **`3` НЕ включается** (вернутся строки 0, 1, 2).
* **`df.loc[строки, столбцы]`** — индексация **по меткам индекса и названиям колонок (Label Location)**.
  * **Внимание!** В срезах по меткам `df.loc["a":"c"]` правая граница **`"c"` ВКЛЮЧАЕТСЯ**!
  * Пример: `df.loc[["id_1", "id_3"], ["price", "qty"]]`.

### 📚 Связанные методы в копилку:
* `df["col"]` — возвращает одну колонку как `Series`. `df[["col1", "col2"]]` — возвращает несколько колонок как `DataFrame`.
* `df.at[row_label, col_label]` и `df.iat[row_pos, col_pos]` — сверхбыстрый доступ к **одному скалярному значению** (быстрее `.loc`/`.iloc`).

### Задача 2
Напиши функцию `extract_subsets(df)`, которая принимает `DataFrame` `df` и возвращает кортеж из двух элементов `(first_three_rows, specific_cells)`:
1. `first_three_rows` — первые 3 строки и первые 2 столбца таблицы, выбранные по их **числовой позиции** (используй `.iloc`).
2. `specific_cells` — подтаблица для строк с метками индекса `"ord_2"` и `"ord_4"` и только для колонок `"item"` и `"price"` (используй `.loc`).

```python
def extract_subsets(df: pd.DataFrame):
    # ТВОЙ КОД ЗДЕСЬ
    return (df.iloc[:3,:2], df.loc[['ord_2','ord_4'],['item','price']])
    # pass


# === ТЕСТЫ ===
test_df = pd.DataFrame(
    {
        "item": ["Laptop", "Mouse", "Keyboard", "Monitor"],
        "price": [1000, 25, 75, 300],
        "stock": [5, 50, 20, 10]
    },
    index=["ord_1", "ord_2", "ord_3", "ord_4"]
)
p1, p2 = extract_subsets(test_df)
assert p1.shape == (3, 2), f"Неверный размер первой выборки: {p1.shape}"
assert list(p2.index) == ["ord_2", "ord_4"] and list(p2.columns) == ["item", "price"], "Ошибка во второй выборке!"
assert p2.loc["ord_4", "price"] == 300, "Неверное значение в specific_cells!"
print("✅ Задача 2 решена верно!")
```

## Модуль 3. Булева фильтрация, `.isin()` и `.between()`

### 🎯 Операции и методы, которые нужны для задачи:
* Как и в NumPy, условия комбинируются через `&` (И), `|` (ИЛИ), `~` (НЕ), а каждое условие **обязательно берется в круглые скобки**:
  `df[(df["age"] > 18) & (df["city"] == "Moscow")]`
* `df["col"].isin([val1, val2, ...])` — возвращает `True` для строк, где значение колонки входит в переданный список (заменяет длинные цепочки `|`).
* `df["col"].between(left, right, inclusive="both")` — проверяет, что число лежит в диапазоне `[left, right]` (по умолчанию обе границы включены).

### 📚 Связанные методы в копилку:
* `df.query("age > 18 and city in ['Moscow', 'SPb']")` — позволяет писать условия фильтрации в виде обычной строки (очень читаемо для длинных условий!). Если нужно сослаться на внешнюю переменную Python внутри строки `query`, перед ней ставится символ `@`: `df.query("price > @min_price")`.

### Задача 3
Напиши функцию `filter_products(df)`, которая принимает таблицу товаров `df` и возвращает отфильтрованный `DataFrame`, где одновременно выполняются три условия:
1. Категория (`"category"`) — это либо `"Electronics"`, либо `"Books"` (используй `.isin()`).
2. Цена (`"price"`) находится в диапазоне от `100` до `500` **включительно** (используй `.between()`).
3. Рейтинг (`"rating"`) **строго больше** `4.0`.

```python
def filter_products(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df[df['category'].isin(['Electronics', "Books"])]
    df = df[df['price'].between(100,500)]
    df = df[df['rating'] > 4.0]
    return df
    # pass


# === ТЕСТЫ ===
products = pd.DataFrame({
    "category": ["Electronics", "Books", "Clothing", "Electronics", "Books"],
    "price": [250, 100, 300, 600, 450],
    "rating": [4.5, 3.8, 4.9, 4.7, 4.2]
})
filtered = filter_products(products)
assert list(filtered.index) == [0, 4], f"Ожидались строки с индексами [0, 4], получено: {list(filtered.index)}"
print("✅ Задача 3 решена верно!")
```

## Модуль 4. Создание и удаление колонок, векторные вычисления

### 🎯 Методы, которые нужны для задачи:
* **Антипаттерн №1 в Pandas:** никогда не используй циклы `for` или `df.iterrows()` для арифметики по столбцам! Все операции (`+`, `-`, `*`, `/`, `np.where`) между колонками работают векторно и мгновенно.
* Чтобы не мутировать исходный `DataFrame` (и не ловить пресловутый `SettingWithCopyWarning`), сначала делай копию `df = df.copy()` или используй `df.assign(...)`.
* `df.drop(columns=["col1", "col2"])` — удаляет указанные колонки и возвращает новый `DataFrame`.
* `df.rename(columns={"old_name": "new_name"})` — переименовывает колонки.

### 📚 Связанные методы в копилку:
* `df.assign(total = df["price"] * df["qty"])` — функциональный способ добавить новые колонки (возвращает копию таблицы, удобно объединять в цепочки методов method chaining).
* `np.select([cond1, cond2], [val1, val2], default=...)` — отлично работает прямо поверх колонок Pandas для создания категориальных признаков по нескольким условиям.

### Задача 4
Напиши функцию `process_orders(df)`, которая принимает `DataFrame` с колонками `"price"`, `"qty"`, `"discount"` (доля скидки от 0 до 1) и `"temp_code"`:
1. Создает копию таблицы (`df = df.copy()`).
2. Добавляет колонку `"final_cost"`, равную `price * qty * (1 - discount)`.
3. Добавляет булеву (или строковую) колонку `"tier"`: если `final_cost >= 1000`, там должно быть значение `"VIP"`, иначе `"Regular"` (используй `np.where`).
4. Удаляет техническую колонку `"temp_code"` и возвращает результат.

```python
def process_orders(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.copy()
    df['final_cost'] = df['price'] * df['qty'] * (1 - df['discount'])
    df['tier'] = np.where(df["final_cost"] >= 1000, "VIP", "Regular")
    df = df.drop(columns=['temp_code'])
    return df
    # pass


# === ТЕСТЫ ===
orders = pd.DataFrame({
    "price": [500.0, 200.0],
    "qty": [3, 2],
    "discount": [0.2, 0.0],
    "temp_code": ["x1", "x2"]
})
processed = process_orders(orders)
assert "temp_code" not in processed.columns, "Колонка temp_code не была удалена!"
assert np.allclose(processed["final_cost"], [1200.0, 400.0]), f"Неверный расчет final_cost: {processed['final_cost'].tolist()}"
assert list(processed["tier"]) == ["VIP", "Regular"], f"Неверный расчет tier: {processed['tier'].tolist()}"
print("✅ Задача 4 решена верно!")
```

## Модуль 5. Преобразование значений (`.map`) и строковые методы (`.str`)

### 🎯 Методы, которые нужны для задачи:
* `df["col"].map(mapping_dict)` — заменяет каждое значение колонки по словарю `mapping_dict` (например, `{"M": 0, "F": 1}`). Значения, которых нет в словаре, превратятся в `NaN`.
* Аксессор **`.str`** открывает доступ к векторным строковым методам сразу для всего столбца:
  * `df["col"].str.strip()` — убирает пробелы по краям.
  * `df["col"].str.lower()` / `.str.upper()` / `.str.title()` — меняет регистр.
  * `df["col"].str.contains("pattern", case=False, na=False)` — проверяет наличие подстроки или регулярного выражения.
  * `df["col"].str.split("@").str[1]` — разбивает строку по разделителю и берет нужный элемент по индексу!

### 📚 Связанные методы в копилку:
* `df["col"].replace(mapping_dict)` — в отличие от `.map()`, оставляет нетронутыми те значения, которых нет в словаре!
* `df["col"].apply(func)` — применяет произвольную Python-функцию к каждому элементу. Работает как цикл под капотом (медленно на миллионах строк), поэтому используй `.apply` только тогда, когда нет встроенного векторного или `.str` метода.

### Задача 5
Напиши функцию `clean_user_profiles(df)`, которая принимает `DataFrame` с колонками `"email"` и `"plan"`:
1. Создает копию таблицы.
2. Из колонки `"email"` (где могут быть лишние пробелы по краям и разный регистр, например `"  User@GMAIL.com "`) извлекает **домен почты** (всё, что после `@`), переведенный в нижний регистр и очищенный от пробелов, и сохраняет в новую колонку `"domain"`.
3. Колонку `"plan"` (`"free"`, `"pro"`, `"enterprise"`) переводит в числовой код с помощью `.map()` по словарю `{"free": 0, "pro": 1, "enterprise": 2}` и сохраняет в колонку `"plan_code"`.
4. Возвращает обновленный `DataFrame`.

```python
def clean_user_profiles(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.copy()
    df['domain'] = df['email'].str.strip().str.lower().str.split('@').str[1]
    df['plan_code'] = df['plan'].map({'free' : 0, "pro" : 1, 'enterprise' : 2})
    return df
    # pass


# === ТЕСТЫ ===
profiles = pd.DataFrame({
    "email": ["  ivan@YANDEX.ru ", "anna@Gmail.com", " boss@corp.org  "],
    "plan": ["free", "pro", "enterprise"]
})
cleaned_prof = clean_user_profiles(profiles)
assert list(cleaned_prof["domain"]) == ["yandex.ru", "gmail.com", "corp.org"], f"Неверные домены: {list(cleaned_prof['domain'])}"
assert list(cleaned_prof["plan_code"]) == [0, 1, 2], f"Неверные коды тарифов: {list(cleaned_prof['plan_code'])}"
print("✅ Задача 5 решена верно!")
```

## Модуль 6. Обработка пропусков (`isna`, `fillna`, `dropna`)

### 🎯 Методы, которые нужны для задачи:
* `df.isna()` (или `df.isnull()`) — возвращает булеву таблицу (`True` там, где пропуск `NaN`, `None` или `pd.NA`).
  * `df.isna().sum()` — считает количество пропусков в каждой колонке.
* `df.dropna(subset=["col1", "col2"])` — удаляет строки, в которых есть пропуск хотя бы в одной из указанных в `subset` колонок.
* `df["col"].fillna(value)` — заполняет пропуски указанным значением (константой, средним `df["col"].mean()`, медианой `df["col"].median()` или словарем значений для разных колонок).

### 📚 Связанные методы в копилку:
* `df.notna()` — обратная маска (`True` там, где значение НЕ пропущено).
* `df["col"].ffill()` / `df["col"].bfill()` — заполняет пропуски предыдущим (`forward fill`) или следующим (`backward fill`) известным значением (постоянно используется во временных рядах!).

### Задача 6
Напиши функцию `handle_missing_values(df)`, которая принимает `DataFrame` с колонками `"user_id"`, `"age"` и `"city"`:
1. Сначала удаляет все строки, где пропущен `"user_id"` (используй `dropna(subset=...)`), и делает `.copy()`.
2. В оставшихся строках заполняет пропуски в колонке `"age"` **медианой** этой колонки (`df["age"].median()`).
3. Заполняет пропуски в колонке `"city"` строкой `"Unknown"`.
4. Возвращает очищенный `DataFrame`.

```python
def handle_missing_values(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.dropna(subset=['user_id']).copy()
    df['age'] = df['age'].fillna(df['age'].median())
    df['city'] = df['city'].fillna("Unknown")
    return df
    # pass


# === ТЕСТЫ ===
raw_df = pd.DataFrame({
    "user_id": [1.0, np.nan, 3.0, 4.0],
    "age": [20.0, 99.0, np.nan, 30.0],
    "city": ["Moscow", "SPb", None, "Kazan"]
})
clean_df = handle_missing_values(raw_df)
assert len(clean_df) == 3, f"Строка с пропущенным user_id должна быть удалена! Длина: {len(clean_df)}"
assert clean_df.loc[2, "age"] == 25.0, f"Медиана возраста [20, 30] равна 25.0, получено: {clean_df.loc[2, 'age']}"
assert clean_df.loc[2, "city"] == "Unknown", f"Пропуск в city должен стать 'Unknown', получено: {clean_df.loc[2, 'city']}"
print("✅ Задача 6 решена верно!")
```

## Модуль 7. Сортировка (`sort_values`) и быстрый поиск Top-N (`nlargest`)

### 🎯 Методы, которые нужны для задачи:
* `df.sort_values(by=["col1", "col2"], ascending=[True, False])` — сортирует таблицу по одной или нескольким колонкам. В `ascending` можно передать список булевых значений, чтобы одну колонку отсортировать по возрастанию, а вторую (при совпадении первой) — по убыванию!
* `df.nlargest(n, "col")` / `df.nsmallest(n, "col")` — возвращает топ-`n` самых больших / маленьких строк по колонке `"col"` без полной сортировки всей таблицы (работает быстрее `sort_values().head(n)`).

### 📚 Связанные методы в копилку:
* `df.sort_index()` — сортирует строки по значениям индекса.
* Параметр `na_position="last"` (по умолчанию в `sort_values`) всегда отправляет `NaN` в самый конец таблицы независимо от направления сортировки.

### Задача 7
Напиши функцию `sort_employees(df)`, которая принимает таблицу сотрудников с колонками `"department"`, `"salary"` и `"name"`:
1. Сортирует сотрудников сначала по отделу (`"department"`) **по алфавиту (возрастанию)**, а внутри одного отдела — по зарплате (`"salary"`) **по убыванию**.
2. Сбрасывает индекс отсортированной таблицы через `.reset_index(drop=True)` и возвращает результат.

```python
def sort_employees(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.sort_values(by=['department','salary'],ascending=[True,False])
    df = df.reset_index(drop=True)
    return df
    # pass


# === ТЕСТЫ ===
emp = pd.DataFrame({
    "department": ["IT", "HR", "IT", "HR"],
    "salary": [150, 90, 200, 110],
    "name": ["Ivan", "Olga", "Petr", "Anna"]
})
sorted_emp = sort_employees(emp)
assert list(sorted_emp["name"]) == ["Anna", "Olga", "Petr", "Ivan"], f"Неверный порядок: {list(sorted_emp['name'])}"
assert list(sorted_emp.index) == [0, 1, 2, 3], "Индекс не был сброшен через reset_index(drop=True)!"
print("✅ Задача 7 решена верно!")
```

## Модуль 8. Поиск и удаление дубликатов (`duplicated`, `drop_duplicates`)

### 🎯 Методы, которые нужны для задачи:
* `df.duplicated(subset=..., keep="first")` — возвращает булеву маску (`True` для строк-дубликатов).
* `df.drop_duplicates(subset=["col1", ...], keep="first")` — удаляет дубликаты по указанным в `subset` колонкам (если `subset` не указан, сравниваются вообще все колонки строки).
  * `keep="first"` (по умолчанию) — оставляет первое вхождение.
  * `keep="last"` — оставляет последнее вхождение (удобно, когда события отсортированы по времени и нам нужно оставить только актуальный статус пользователя!).
  * `keep=False` — удаляет **все** дублирующиеся строки целиком.

### 📚 Классический паттерн с собеседований:
Как оставить только **последнюю транзакцию каждого пользователя**? Сначала делаем `df.sort_values("timestamp")`, а затем `df.drop_duplicates(subset=["user_id"], keep="last")`!

### Задача 8
В логах событий `df` есть колонки `"user_id"`, `"timestamp"` (целое число) и `"status"`. События могут идти вразнобой.

Напиши функцию `get_latest_user_status(df)`, которая для каждого `"user_id"` оставляет только **одну запись с самым большим `"timestamp"`** (последнее по времени состояние), отсортированную по `"user_id"` по возрастанию.

```python
def get_latest_user_status(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.sort_values(by=['user_id','timestamp'])
    df = df.drop_duplicates(subset=['user_id'],keep='last')
    return df
    # pass


# === ТЕСТЫ ===
logs = pd.DataFrame({
    "user_id": [2, 1, 2, 1],
    "timestamp": [10, 50, 30, 20],
    "status": ["login", "purchase", "logout", "login"]
})
latest = get_latest_user_status(logs)
assert len(latest) == 2, f"Должно остаться 2 пользователя, получено: {len(latest)}"
assert list(latest["user_id"]) == [1, 2], "Таблица должна быть отсортирована по user_id"
assert list(latest["status"]) == ["purchase", "logout"], f"Неверные последние статусы: {list(latest['status'])}"
print("✅ Задача 8 решена верно!")
```

## Модуль 9. Частотный анализ (`value_counts`, `nunique`) и описательные статистики

### 🎯 Методы, которые нужны для задачи:
* `df["col"].nunique(dropna=True)` — возвращает количество **уникальных** значений в колонке (`int`).
* `df["col"].unique()` — возвращает массив самих уникальных значений.
* `df["col"].value_counts(normalize=False, dropna=True)` — считает частоту встречаемости каждого значения и возвращает `Series`, отсортированный по убыванию частоты.
  * Если передать **`normalize=True`**, вместо абсолютных чисел вернет **доли (вероятности)** от `0.0` до `1.0`!

### 📚 Связанные методы в копилку:
* `df.describe()` — сводная таблица базовых статистик (`count`, `mean`, `std`, `min`, квартили `25%`, `50%`, `75%`, `max`) по всем числовым колонкам.
* `df.corr(numeric_only=True)` — матрица попарных корреляций Пирсона между столбцами.
* `df["col"].idxmax()` / `df["col"].idxmin()` — возвращает **метку индекса** строки, где достигается максимум / минимум (аналог `argmax`, но возвращает именно `index`).

### Задача 9
Напиши функцию `analyze_categories(df)`, которая принимает `DataFrame` с колонкой `"brand"` и возвращает кортеж из трех элементов `(num_unique_brands, most_frequent_brand, top_brand_share)`:
1. `num_unique_brands` — количество уникальных брендов в колонке `"brand"` (`int`).
2. `most_frequent_brand` — название самого часто встречающегося бренда (`str`).
3. `top_brand_share` — доля (от `0.0` до `1.0`, тип `float`) самого популярного бренда среди всех непустых записей (используй `value_counts(normalize=True)`).

```python
def analyze_categories(df: pd.DataFrame):
    # ТВОЙ КОД ЗДЕСЬ
    return (df['brand'].nunique(dropna=True), df['brand'].value_counts(dropna=True).index[0], df['brand'].value_counts(normalize=True,dropna=True).iloc[0])
    # pass


# === ТЕСТЫ ===
brands_df = pd.DataFrame({
    "brand": ["Apple", "Samsung", "Apple", "Xiaomi", "Apple", "Samsung", "Apple", "Xiaomi"]
})
n_uniq, top_b, share = analyze_categories(brands_df)
assert n_uniq == 3, f"Ожидалось 3 уникальных бренда, получено: {n_uniq}"
assert top_b == "Apple", f"Самый частый бренд — Apple, получено: {top_b}"
assert np.isclose(share, 0.5), f"Доля Apple (4 из 8) равна 0.5, получено: {share}"
print("✅ Задача 9 решена верно!")
```

## Модуль 10. Приведение типов (`astype`, `pd.to_numeric`) и биннинг (`pd.cut`)

### 🎯 Методы, которые нужны для задачи:
* `pd.to_numeric(df["col"], errors="coerce")` — безопасно переводит строковую колонку в числа (`float`/`int`). Если попадается битая строка (например, `"error"` или `"-"`), параметр **`errors="coerce"`** не роняет код с ошибкой `ValueError`, а аккуратно превращает битое значение в `NaN`!
* `df["col"].astype(dtype)` — прямое приведение типа (например, `int`, `float`, `str`, `"category"`).
* `pd.cut(df["col"], bins=[b0, b1, b2, ...], labels=[...])` — разбивает непрерывную числовую колонку на интервалы (корзины) и присваивает им категории `labels`.

### 📚 Связанные методы в копилку:
* `pd.qcut(df["col"], q=4, labels=...)` — разбивает значения на корзины **равного размера по квантилям** (например, на 4 квартиля с одинаковым числом объектов в каждой группе).
* Тип `"category"` (`df["col"].astype("category")`) сжимает строковые колонки с небольшим числом уникальных значений в памяти в 5–10 раз и ускоряет `groupby`.

### Задача 10
Сырые данные часто приходят со строковым мусором внутри числовых полей.

Напиши функцию `clean_and_bin_ages(df)`, которая принимает `DataFrame` с колонкой `"raw_age"` (строки, среди которых могут быть текстовые опечатки вроде `"unknown"` или `"N/A"`):
1. Создает копию `df = df.copy()`.
2. Конвертирует `"raw_age"` в числа с помощью `pd.to_numeric(..., errors="coerce")` и сохраняет в колонку `"age"`.
3. Удаляет строки, где `"age"` оказался `NaN`.
4. Создает колонку `"age_group"` с помощью `pd.cut`, разбивая `"age"` по границам `bins=[0, 18, 60, 120]` с метками `labels=["child", "adult", "senior"]`.
5. Возвращает полученный `DataFrame`.

```python
def clean_and_bin_ages(df: pd.DataFrame) -> pd.DataFrame:
    # ТВОЙ КОД ЗДЕСЬ
    df = df.copy()
    df['age'] = pd.to_numeric(df['raw_age'], errors='coerce')
    df = df.dropna(subset=['age'])
    df['age_group'] = pd.cut(df['age'],bins=[0,18,60,120],labels=['child','adult','senior'])
    return df
    # pass


# === ТЕСТЫ ===
dirty = pd.DataFrame({
    "raw_age": ["15", "25", "unknown", "70", "bad_val"]
})
res = clean_and_bin_ages(dirty)
assert len(res) == 3, f"Должно остаться 3 валидные строки, получено: {len(res)}"
assert list(res["age"]) == [15.0, 25.0, 70.0], f"Неверные числовые значения age: {list(res['age'])}"
assert list(res["age_group"]) == ["child", "adult", "senior"], f"Неверные возрастные группы: {list(res['age_group'])}"
print("🎉 ПОЗДРАВЛЯЮ! Уровень 1 по Pandas (10 из 10) полностью пройден!")
```
